```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a one-qubit circuit
qc = QuantumCircuit(1)

# Apply S gate
qc.s(0)

# Obtain the statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit")
print(qc)

print("\nStatevector")
print(state)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

# Create circuit
qc = QuantumCircuit(1)

# Apply Hadamard then T gate
qc.h(0)
qc.t(0)

# Generate statevector
sv = Statevector.from_instruction(qc)

print(qc)
print("\nStatevector")
print(sv)

plot_bloch_multivector(sv)
plt.show()
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create two-qubit circuit
qc = QuantumCircuit(2)

# Apply different gates
qc.x(1)
qc.h(0)
qc.cx(0, 1)

# Generate statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit")
print(qc)

print("\nBasis State Amplitudes")

for i, amplitude in enumerate(state.data):
    print(f"|{i:02b}> : {amplitude}")
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Create circuit
qc = QuantumCircuit(2)

qc.h(0)
qc.y(1)

# Generate statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit")
print(qc)

print("\nStatevector")
print(state)

prob = np.abs(state.data) ** 2

print("\nProbability of Each Basis State")

for i, p in enumerate(prob):
    print(f"|{i:02b}> : {p:.4f}")

print("\nTotal Probability =", np.sum(prob))
```
