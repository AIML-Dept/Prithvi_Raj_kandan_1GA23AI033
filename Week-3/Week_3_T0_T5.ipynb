```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

circuits = {
    "Pauli-X": QuantumCircuit(1),
    "Pauli-Y": QuantumCircuit(1),
    "Pauli-Z": QuantumCircuit(1)
}

circuits["Pauli-X"].x(0)
circuits["Pauli-Y"].y(0)
circuits["Pauli-Z"].z(0)

for name, qc in circuits.items():
    state = Statevector.from_instruction(qc)
    print(f"\n{name} Gate")
    print(qc)
    print("Statevector:", state)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)

final_state = Statevector.from_instruction(qc)

print("Circuit:")
print(qc)

print("\nFinal State:")
print(final_state)

print("\nProbabilities:")
print(final_state.probabilities())
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

gate_sequence = [
    qc.t,
    qc.h,
    qc.z,
    qc.s,
    qc.y
]

for gate in gate_sequence:
    gate(0)

state = Statevector.from_instruction(qc)

print("Quantum Circuit")
print(qc)

print("\nStatevector")
print(state)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

coin = QuantumCircuit(1)

coin.h(0)

result = Statevector.from_instruction(coin)

head, tail = result.probabilities()

print("Quantum Coin Flip")
print(coin)

print(f"\nHead Probability : {head:.3f}")
print(f"Tail Probability : {tail:.3f}")

if head == tail:
    print("\nThe coin is fair.")
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# HZH Circuit
qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.z(0)
qc1.h(0)

# X Circuit
qc2 = QuantumCircuit(1)
qc2.x(0)

state1 = Statevector.from_instruction(qc1)
state2 = Statevector.from_instruction(qc2)

print("Circuit 1 (HZH)")
print(qc1)

print("\nCircuit 2 (X)")
print(qc2)

print("\nAre the two statevectors equivalent?")
print(state1.equiv(state2))
```
