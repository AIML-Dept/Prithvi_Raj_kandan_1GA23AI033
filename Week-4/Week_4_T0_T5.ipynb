```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

circuit = QuantumCircuit(2)

# Prepare control qubit as 1
circuit.x(0)

# Controlled-X operation
circuit.cx(control_qubit=0, target_qubit=1)

output = Statevector.from_instruction(circuit)

print("=== CNOT Experiment ===")
print(circuit.draw())

print("\nResulting State:")
print(output)

print("\nProbability:")
print(output.probabilities_dict())
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

bell_circuit = QuantumCircuit(2, 2)

# First qubit enters superposition
bell_circuit.h(0)

# Create entanglement
bell_circuit.cx(0, 1)

# Measurement
for qubit in range(2):
    bell_circuit.measure(qubit, qubit)

backend = AerSimulator()

simulation = backend.run(
    bell_circuit,
    shots=1024
)

result = simulation.result()
frequency = result.get_counts()

print("=== Bell State ===")
print(bell_circuit.draw())

print("\nMeasurement Frequency:")
for state, count in sorted(frequency.items()):
    print(state, ":", count)
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

ghz_circuit = QuantumCircuit(3, 3)

# Start the GHZ preparation
ghz_circuit.h(0)

# Spread the correlation
ghz_circuit.cx(0, 1)
ghz_circuit.cx(0, 2)

# Measure every qubit
ghz_circuit.measure_all()

machine = AerSimulator()

result = machine.run(
    ghz_circuit,
    shots=1024
).result()

values = result.get_counts()

print("=== GHZ State ===")
print(ghz_circuit.draw())

print("\nResults:")
print(values)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def build_full_adder():
    circuit = QuantumCircuit(5)

    # Inputs:
    # q0 = A
    # q1 = B
    # q2 = Carry-in
    #
    # Outputs:
    # q3 = Sum
    # q4 = Carry

    # Generate SUM = A XOR B XOR Cin
    for source in [0, 1, 2]:
        circuit.cx(source, 3)

    # Generate CARRY
    pairs = [(0, 1), (0, 2), (1, 2)]

    for first, second in pairs:
        circuit.ccx(first, second, 4)

    return circuit


full_adder = build_full_adder()

answer = Statevector.from_instruction(full_adder)

print("=== Reversible Full Adder ===")
print(full_adder.draw())

print("\nStatevector:")
print(answer)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# HZH Circuit
qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.z(0)
qc1.h(0)

# X Circuit
qc2 = QuantumCircuit(1)
qc2.x(0)

state1 = Statevector.from_instruction(qc1)
state2 = Statevector.from_instruction(qc2)

print("Circuit 1 (HZH)")
print(qc1)

print("\nCircuit 2 (X)")
print(qc2)

print("\nAre the two statevectors equivalent?")
print(state1.equiv(state2))
```
